```python
import pandas as pd
customers = pd.DataFrame({
    "Customer_ID": [1, 2, 3, 4, 5],
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Arjun"],
    "City": ["Mumbai", "Pune", "Delhi", "Hyderabad", "Chennai"]
})
orders = pd.DataFrame({
    "Customer_ID": [1, 2, 3, 6],
    "Product": ["Laptop", "Phone", "Tablet", "Mouse"],
    "Amount": [50000, 20000, 30000, 1500]
})
print("Customers DataFrame:")
print(customers)
print("\nOrders DataFrame:")
print(orders)
```

```python
inner_join = pd.merge(customers, orders, on="Customer_ID", how="inner")
print(inner_join)
```

```python
left_join = pd.merge(customers, orders, on="Customer_ID", how="left")
print(left_join)
```

```python
right_join = pd.merge(customers, orders, on="Customer_ID", how="right")
print(right_join)
```

```python
outer_join = pd.merge(customers, orders, on="Customer_ID", how="outer")
print(outer_join)
```

```python
selected_merge = pd.merge(customers[["Customer_ID", "Name"]], orders[["Customer_ID", "Amount"]], on="Customer_ID", how="inner")
print(selected_merge)
```

```python
customers_renamed = customers.rename(columns={"Customer_ID": "ID"})
merged_data = pd.merge(customers_renamed, orders, left_on="ID", right_on="Customer_ID", how="inner")
print(merged_data)
```

```python
customers_index = customers.set_index("Customer_ID")
orders_index = orders.set_index("Customer_ID")
joined_data = customers_index.join(orders_index, how="inner", lsuffix="_customer", rsuffix="_order")
print(joined_data)
```

```python
left_join_index = customers_index.join(orders_index, how="left", lsuffix="_customer", rsuffix="_order")
print(left_join_index)
```

```python
combined_data = pd.concat([customers, customers], ignore_index=True)
print(combined_data)
```

```python
horizontal_data = pd.concat([customers, orders], axis=1)
print(horizontal_data)
```

```python
print("Customers DataFrame:")
print(customers)
print("\nOrders DataFrame:")
print(orders)
print("\nFinal Merged DataFrame:")
print(pd.merge(customers, orders, on="Customer_ID", how="outer"))
```
